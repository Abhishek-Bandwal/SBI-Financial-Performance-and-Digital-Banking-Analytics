
## 13. Separation of Identifier and KPI Columns

The dataset contains two types of information:

**Identifier columns**

- ID

These columns identify the bank and financial year associated with each observation.

**Financial KPI columns**

The remaining 23 columns contain the financial measures used to understand bank performance.

```python
identifier_columns = ["ID", "Bank", "Financial Year"]

kpi_columns = df.columns.drop(identifier_columns)

print("Number of identifier columns:", len(identifier_columns))
print("Number of KPI columns:", len(kpi_columns))

print("\nKPI columns:\n")
print(kpi_columns.tolist())
```


## 14. Financial KPI Classification

The 23 financial KPIs are grouped into six business-oriented categories to make the dataset easier to understand and organise.

### 14.1 Business Size — 3 KPIs

- Total Assets
- Total Deposits
- Gross Advances

These indicators represent the overall scale of a bank's financial operations.

### 14.2 Profitability — 5 KPIs

- Net Interest Income (NII)
- Operating Profit
- Net Profit (PAT)
- Net Interest Margin (NIM)
- Return on Assets (ROA)

These indicators represent income generation and profitability.

### 14.3 Asset Quality — 3 KPIs

- Gross NPA (GNPA) Ratio
- Net NPA (NNPA) Ratio
- Provision Coverage Ratio (PCR)

These indicators provide information about asset quality and provisioning.

### 14.4 Growth — 6 KPIs

- Total Assets Growth
- Deposit Growth
- Gross Advances Growth
- NII Growth
- Operating Profit Growth
- Net Profit Growth

These indicators represent changes in financial measures over time.

### 14.5 Capital & Liquidity — 3 KPIs

- CASA Ratio
- Capital Adequacy Ratio (CRAR)
- Common Equity Tier 1 (CET1) Ratio

These indicators cover selected funding and capital-strength measures.

### 14.6 Efficiency & Shareholder Value — 3 KPIs

- Cost-to-Income Ratio
- Earnings Per Share (EPS)
- Book Value Per Share

These indicators represent operating efficiency and selected shareholder-value measures.

**Result:** All 23 KPIs have been organised into six analytical categories.


```python
kpi_categories = {
    "Business Size": [
        "Total Assets (₹ crore)",
        "Total Deposits (₹ crore)",
        "Gross Advances (₹ crore)"
    ],

    "Profitability": [
        "Net Interest Income (NII) (₹ crore)",
        "Operating Profit (₹ crore)",
        "Net Profit (PAT) (₹ crore)",
        "Net Interest Margin (NIM) (%)",
        "Return on Assets (ROA) (%)"
    ],

    "Asset Quality": [
        "Gross NPA (GNPA) Ratio (%)",
        "Net NPA (NNPA) Ratio (%)",
        "Provision Coverage Ratio (PCR) (%)"
    ],

    "Growth": [
        "Total Assets Growth (%)",
        "Deposit Growth (%)",
        "Gross Advances Growth (%)",
        "NII Growth (%)",
        "Operating Profit Growth (%)",
        "Net Profit Growth (%)"
    ],

    "Capital & Liquidity": [
        "CASA Ratio (%)",
        "Capital Adequacy Ratio (CRAR) (%)",
        "Common Equity Tier 1 (CET1) Ratio (%)"
    ],

    "Efficiency & Shareholder Value": [
        "Cost-to-Income Ratio (%)",
        "Earnings Per Share (EPS) (₹/share)",
        "Book Value Per Share (₹/share)"
    ]
}

for category, kpis in kpi_categories.items():
    print(f"\n{category}: {len(kpis)} KPIs")
    for kpi in kpis:
        print("-", kpi)
```


## 15. KPI Classification Validation

After assigning the KPIs to their respective categories, a validation check is performed to confirm that:

- All 23 KPI columns have been classified.
- No KPI has been omitted.
- No KPI has been assigned to multiple categories.

**Observation:**

- Total classified KPIs: 23
- Unique classified KPIs: 23
- Unclassified KPIs: 0
- KPIs classified more than once: 0

**Result:** The classification is complete, with every KPI assigned exactly once.


```python
classified_kpis = []

for kpis in kpi_categories.values():
    classified_kpis.extend(kpis)

print("Total classified KPIs:", len(classified_kpis))
print("Unique classified KPIs:", len(set(classified_kpis)))

print("\nKPIs not classified:")
print(set(kpi_columns) - set(classified_kpis))

print("\nKPIs classified more than once:")
print([
    kpi for kpi in set(classified_kpis)
    if classified_kpis.count(kpi) > 1
])
```


## 16. KPI Measurement Units

The measurement units of the financial KPIs are examined to understand how each indicator is represented.

The KPIs fall into three measurement types:

| Measurement Unit | Number of KPIs |
|---|---:|
| ₹ crore | 6 |
| Percentage | 15 |
| ₹ per share | 2 |
| **Total** | **23** |

**Observation:**

- Financial amounts such as Total Assets, Deposits and Net Profit are measured in ₹ crore.
- Ratios and growth indicators are measured in percentages.
- EPS and Book Value Per Share are measured in ₹ per share.

**Important:** The growth KPIs are stored as decimal proportions. For example, a stored value of `0.1061` represents a growth rate of approximately `10.61%`.

Understanding these formats is important when interpreting the statistical results and preparing future visualisations.


```python
for category, kpis in kpi_categories.items():

    print(f"\n{category}")

    for kpi in kpis:

        if "(₹ crore)" in kpi:
            unit = "₹ crore"

        elif "(₹/share)" in kpi:
            unit = "₹ per share"

        elif "(%)" in kpi:
            unit = "Percentage"

        else:
            unit = "Unit not identified"

        print(f"{kpi} : {unit}")
```
